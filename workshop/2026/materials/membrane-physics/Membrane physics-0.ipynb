```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LightSource
from ipywidgets import interact, FloatSlider

# 解决中文显示与负号
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

def plot_curvature_surface(c1=1.0, c2=1.0):
    """
    通过两个主曲率 c1 和 c2 拟合局部二次曲面 z = 0.5 * (c1 * x^2 + c2 * y^2)
    计算并展示平均曲率 H 和高斯曲率 K 的几何含义
    """
    # 建立局部 x, y 网格
    x = np.linspace(-1, 1, 40)
    y = np.linspace(-1, 1, 40)
    X, Y = np.meshgrid(x, y)
    
    # 局部二次曲面公式 (Taylor 展开表达)
    Z = 0.5 * (c1 * X**2 + c2 * Y**2)
    
    # 计算平均曲率 H 与高斯曲率 K (原点处)
    H = 0.5 * (c1 + c2)
    K = c1 * c2

    # 创建画布
    fig = plt.figure(figsize=(12, 5), facecolor='#f8f9fa')
    
    # 1. 3D 可视化曲面
    ax1 = fig.add_subplot(121, projection='3d', facecolor='#f8f9fa')
    ls = LightSource(azdeg=135, altdeg=45)
    
    # 用 Z 轴高度作为色彩度
    rgb = ls.shade(Z, cmap=plt.cm.coolwarm, vert_exag=0.1, blend_mode='soft')
    
    ax1.plot_surface(X, Y, Z, facecolors=rgb, rstride=1, cstride=1, linewidth=0, antialiased=True)
    ax1.set_xlim([-1, 1])
    ax1.set_ylim([-1, 1])
    ax1.set_zlim([-1.5, 1.5])
    ax1.view_init(elev=20, azim=45)
    ax1.axis('off')
    
    # 标题展示实时计算的物理量
    ax1.set_title(f"3D 局部曲面\n主曲率 c1={c1:.2f}, c2={c2:.2f}", fontsize=12, fontweight='bold')
    
    # 2. 2D 数值与曲率类型判断面板
    ax2 = fig.add_subplot(122)
    ax2.axis('off')
    
    # 判断曲面几何类型
    if K > 0.05:
        shape_type = "椭圆点 (Elliptic Point)\n例如：球面 / 凸囊泡"
        color = "#2a9d8f"
    elif K < -0.05:
        shape_type = "马鞍点 / 双曲点 (Hyperbolic Point)\n例如：马鞍面 / 膜出芽颈部"
        color = "#e76f51"
    else:
        shape_type = "抛物点 / 抛物线 (Parabolic Point)\n例如：圆柱面 / 膜成管"
        color = "#f4a261"

    # 展示数学文本
    info_text = (
        f"【实时几何量计算】\n\n"
        f"• 第一主曲率 c1 = {c1:.2f}\n"
        f"• 第二主曲率 c2 = {c2:.2f}\n\n"
        f"-----------------------------------\n"
        f"• 平均曲率 H = (c1 + c2) / 2 = {H:.2f}\n"
        f"• 高斯曲率 K = c1 * c2       = {K:.2f}\n"
        f"-----------------------------------\n\n"
        f"【几何构型分类】：\n{shape_type}"
    )
    
    ax2.text(0.1, 0.5, info_text, fontsize=13, verticalalignment='center',
             bbox=dict(boxstyle="round,pad=0.8", facecolor=color, alpha=0.15, edgecolor=color, lw=2))

    plt.tight_layout()
    plt.show()

# 创建滑块交互
interact(
    plot_curvature_surface,
    c1=FloatSlider(min=-2.0, max=2.0, step=0.1, value=1.0, continuous_update=False, description='主曲率 c1'),
    c2=FloatSlider(min=-2.0, max=2.0, step=0.1, value=1.0, continuous_update=False, description='主曲率 c2')
);
```
