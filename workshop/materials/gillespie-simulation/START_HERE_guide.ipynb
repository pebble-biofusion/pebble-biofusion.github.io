## Cheat sheet — everything on one page

**The two random numbers** (state $\mathbf{x}$, propensities $a_j$, total $a_0=\sum_j a_j$):

$$r_1\sim U(0,1):\quad \tau=\frac{\ln(1/r_1)}{a_0}\qquad\text{(when)}\qquad\qquad
r_2\sim U(0,1):\quad j=\text{first index with }\sum_{k\le j}a_k>r_2\,a_0\qquad\text{(which)}$$

**One algorithm step:** compute $a_j$ and $a_0$ $\to$ draw $r_1,r_2$ $\to$ get $\tau$ and $j$
$\to$ update $t\leftarrow t+\tau$, $\mathbf{x}\leftarrow\mathbf{x}+\boldsymbol{\nu}_j$ $\to$ repeat.


## Troubleshooting / FAQ

* **The sliders don't appear / I just see `<interactive ...>`.** Run the notebook **top to bottom**
  with *Run All*; make sure you are in JupyterLab or classic Jupyter Notebook (not a static preview).
  If still missing, install widgets: `pip install ipywidgets`, then `jupyter nbextension enable
  --py widgetsnbextension` (classic) — in JupyterLab it usually just works.
* **A notebook feels slow.** Lower `n_cells` / `Tmax` in module 1, or `Tmax` / `n_reps` in the
  ratio sweep of module 2. SSA slows down when many events fire per unit time (see slide 14).
* **My SinI-SinR model shows no pulses.** Check reaction 7's condition (must fire only when
  $R=0$) and that complexation (`c*I*R`) is large enough to occasionally strip SinR to zero.
* **Where do I get the `gillespie_ssa` function for the case study?** It is defined in
  `gillespie_derivation.ipynb` (last code cell) — open that notebook, run the cell, then copy the
  function into your script, or paste your propensities/stoichiometry into that notebook.
