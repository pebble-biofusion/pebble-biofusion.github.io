
# Multiscale modeling of the cAMP frequency-to-amplitude converter

**Goal:** Use three modules to answer one question: why can high- and low-frequency inputs with the same mean light dose produce different sfGFP outputs?

![FDCC reaction chain from Fig. 1b](assets/fdcc_reaction_scheme_fig1b.png)

$$
\text{blue-light square wave}
\rightarrow \mathrm{bPAC}^{*}
\rightarrow \mathrm{cAMP}
\rightarrow \mathrm{Vfr{-}cAMP}
\rightarrow \mathrm{Activated\ Plac}
\rightarrow \mathrm{mRNA}
\rightarrow \mathrm{sfGFP}.
$$

- **M1: shape the waveform.** High- and low-frequency inputs with the same mean form different cAMP peaks and troughs.
- **M2: rectify nonlinearly.** Molecular binding converts waveform differences into different mean Plac activities.
- **M3: integrate the readout.** Transcription and translation stabilize the difference as protein concentration.



## 1. M1: Same mean, different waveform

Blue light drives cAMP production, while CpdA hydrolyzes cAMP:

$$
\underbrace{\frac{\mathrm d c_{\mathrm{cAMP}}}{\mathrm dt}}
_{\text{cAMP accumulation}}
=
\underbrace{v_{\mathrm{prod,cAMP}}^{\max}u_{\mathrm{light}}(t)}
_{\text{light-driven production}}
-
\underbrace{k_{\mathrm{deg,cAMP}}c_{\mathrm{cAMP}}}
_{\text{CpdA-mediated hydrolysis}}.
$$

At fixed duty cycle $D$, integrate over one cycle at periodic steady state:

$$
\underbrace{0}_{\text{periodic balance}}
=
\underbrace{v_{\mathrm{prod,cAMP}}^{\max}DT}_{\text{production per cycle}}
-
\underbrace{k_{\mathrm{deg,cAMP}}T
\langle c_{\mathrm{cAMP}}\rangle}_{\text{hydrolysis per cycle}}
\quad\Longrightarrow\quad
\underbrace{\langle c_{\mathrm{cAMP}}\rangle}_{\text{cycle-mean cAMP}}
=
\underbrace{\frac{v_{\mathrm{prod,cAMP}}^{\max}}
{k_{\mathrm{deg,cAMP}}}D}_{\text{period-independent mean}}.
$$

**Key result:** Changing the period $T$ preserves cycle-mean cAMP but changes the peaks, troughs, and residence times. Frequency is first encoded in the waveform.


## 2. M2: Rectifying the waveform with a nonlinear boundary

Under the rapid-equilibrium approximation, the two molecular binding stages reduce to a continuous sigmoidal response. Let $\lambda=[\mathrm{Vfr}]_0/K_2$:

$$
\begin{aligned}
\underbrace{\psi(c)}_{\text{composite Plac response}}
&=
\underbrace{\frac{\lambda c^2}
{K_1^2+(\lambda+1)c^2}}_{\text{two-stage molecular nonlinearity}},
\\[4pt]
\underbrace{\theta_P}_{\text{maximum-slope cAMP threshold}}
&=
\underbrace{\frac{K_1}{\sqrt{3(\lambda+1)}}}
_{\text{composite response boundary}}.
\end{aligned}
$$

Nonlinearity means that

$$
\underbrace{\langle\psi(c(t))\rangle}_{\text{mean nonlinear output}}
\neq
\underbrace{\psi(\langle c(t)\rangle)}_{\text{response at the input mean}}.
$$

**Key result:** Even when two cAMP waveforms have the same mean, their peaks and residence times near the threshold can generate different mean Plac activities.


## 3. M3: Integrating and stabilizing the protein readout

M3 is a linear transcription-translation cascade:

$$
\begin{aligned}
\underbrace{\dot c_{\mathrm{mRNA}}}_{\text{mRNA accumulation}}
&=
\underbrace{k_{\mathrm{prod,mRNA}}c_{\mathrm{Plac}^{*}}}
_{\text{Plac-driven transcription}}
-
\underbrace{k_{\mathrm{deg,mRNA}}c_{\mathrm{mRNA}}}
_{\text{mRNA degradation}},
\\[4pt]
\underbrace{\dot c_{\mathrm{sfGFP}}}_{\text{sfGFP accumulation}}
&=
\underbrace{k_{\mathrm{prod,sfGFP}}c_{\mathrm{mRNA}}}
_{\text{mRNA-driven translation}}
-
\underbrace{k_{\mathrm{deg,sfGFP}}c_{\mathrm{sfGFP}}}
_{\text{sfGFP degradation}}.
\end{aligned}
$$

In the linear periodic-steady model, the common gain cancels after normalization:

$$
\underbrace{Y_{\mathrm{sfGFP}}(f)}_{\text{normalized sfGFP response}}
=
\underbrace{\frac{\langle c_{\mathrm{sfGFP}}\rangle_f}
{c_{\mathrm{sfGFP,continuous\ light}}}}_{\text{normalized protein mean}}
=
\underbrace{\frac{\langle c_{\mathrm{Plac}^{*}}\rangle_f}
{c_{\mathrm{Plac}^{*},\mathrm{continuous\ light}}}}
_{\text{normalized promoter response}}.
$$

**Key result:** M3 changes absolute expression, delay, and within-cycle fluctuations, while normalized frequency selection is determined mainly by upstream M1+M2.


## 5. Summary

$$
\text{M1 shapes the waveform}
\;\rightarrow\;
\text{M2 rectifies nonlinearly}
\;\rightarrow\;
\text{M3 integrates and stabilizes the readout}.
$$

No single molecule reads frequency by itself. Three operations at different timescales act together: **frequency is encoded as a waveform, the waveform is rectified into a mean difference, and that difference is accumulated as protein concentration.**
